```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("D:/Datasets/Unemployment in India.csv")
df
```

```python
df.columns = df.columns.str.strip()
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)
```

```python
region_data = df.groupby("Region")["Estimated Unemployment Rate (%)"].mean()

plt.figure(figsize=(12,6))
region_data.sort_values().plot(kind="bar")
plt.title("Average Unemployment Rate by Region")
plt.xlabel("Region")
plt.ylabel("Unemployment Rate (%)")
plt.xticks(rotation=90)
plt.show()
```

```python
area_data = df.groupby("Area")["Estimated Unemployment Rate (%)"].mean()

plt.figure(figsize=(5,5))
area_data.plot(kind="bar")
plt.title("Urban vs Rural Unemployment Rate")
plt.ylabel("Unemployment Rate (%)")
plt.show()
```

```python
monthly = df.groupby("Date")["Estimated Unemployment Rate (%)"].mean()

plt.figure(figsize=(12,5))
plt.plot(monthly.index, monthly.values)
plt.title("Monthly Unemployment Trend")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")
plt.grid(True)
plt.show()
```

```python
print("Highest Unemployment Region:")
print(region_data.idxmax())

print("\nLowest Unemployment Region:")
print(region_data.idxmin())
```
